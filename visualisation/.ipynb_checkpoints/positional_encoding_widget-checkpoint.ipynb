```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display


def generer_positional_encoding(max_seq_len, d_model):
    """Génère la matrice de Positional Encoding."""
    pe = np.zeros((max_seq_len, d_model))
    position = np.arange(0, max_seq_len)[:, np.newaxis]
    div_term = np.exp(np.arange(0, d_model, 2) * -(np.log(10000.0) / d_model))

    pe[:, 0::2] = np.sin(position * div_term)
    pe[:, 1::2] = np.cos(position * div_term)

    return pe


def afficher_graphiques(max_seq_len, d_model):
    """Fonction appelée par le widget pour mettre à jour l'affichage."""
    # On s'assure que d_model est toujours un nombre pair
    if d_model % 2 != 0:
        d_model += 1

    pe_matrix = generer_positional_encoding(max_seq_len, d_model)

    plt.figure(figsize=(16, 6))

    # 1. Heatmap (Matrice globale)
    plt.subplot(1, 2, 1)
    plt.pcolormesh(pe_matrix, cmap="RdBu", vmin=-1, vmax=1)
    plt.title(
        f"Matrice de Positional Encoding\n(d_model={d_model}, seq_len={max_seq_len})"
    )
    plt.xlabel("Dimensions (0 à d_model)")
    plt.ylabel("Position du token")
    plt.colorbar(label="Valeur")
    plt.gca().invert_yaxis()

    # 2. Courbes d'évolution pour les premières dimensions
    plt.subplot(1, 2, 2)

    # On trace les 4 premières dimensions pour voir l'alternance sin/cos
    # et la différence de fréquence. On vérifie que la taille le permet.
    if d_model >= 4:
        plt.plot(
            pe_matrix[:, 0],
            label="Dim 0 (Sinus, très rapide)",
            color="#1f77b4",
            linewidth=2,
        )
        plt.plot(
            pe_matrix[:, 1],
            label="Dim 1 (Cosinus, très rapide)",
            color="#ff7f0e",
            linewidth=2,
            linestyle="--",
        )
        plt.plot(
            pe_matrix[:, 2],
            label="Dim 2 (Sinus, un peu plus lent)",
            color="#2ca02c",
            linewidth=2,
        )
        plt.plot(
            pe_matrix[:, 3],
            label="Dim 3 (Cosinus, un peu plus lent)",
            color="#d62728",
            linewidth=2,
            linestyle="--",
        )

    plt.title("Oscillations des premières dimensions")
    plt.xlabel("Position du token dans la phrase")
    plt.ylabel("Valeur (-1 à 1)")
    plt.legend(loc="upper right")
    plt.grid(True, alpha=0.4)

    plt.tight_layout()
    plt.show()


# --- CRÉATION DES WIDGETS ---
# On crée des curseurs pour manipuler les valeurs en direct
widget_seq_len = widgets.IntSlider(
    value=50,
    min=10,
    max=200,
    step=10,
    description="Nb de Tokens:",
    continuous_update=False,  # Met à jour le graphique seulement quand on lâche le clic
)

widget_d_model = widgets.IntSlider(
    value=64,
    min=16,
    max=512,
    step=16,
    description="Dimensions:",
    continuous_update=False,
)

# On relie la fonction d'affichage aux widgets
ui = widgets.interact(
    afficher_graphiques, max_seq_len=widget_seq_len, d_model=widget_d_model
)
```
